```python
def is_variable(expr):

    return isinstance(expr, str) and expr.islower() and len(expr) == 1

def unify(expr1, expr2, subst=None):
    if subst is None:
        subst = {}

    if expr1 == expr2:
        return subst


    if is_variable(expr1):
        return unify_var(expr1, expr2, subst)
    if is_variable(expr2):
        return unify_var(expr2, expr1, subst)


    if isinstance(expr1, (list, tuple)) and isinstance(expr2, (list, tuple)):
        if len(expr1) != len(expr2):
            return None

        for item1, item2 in zip(expr1, expr2):
            subst = unify(item1, item2, subst)
            if subst is None:
                return None
        return subst

    return None

def unify_var(var, val, subst):
    if var in subst:
        return unify(subst[var], val, subst)
    if val in subst:
        return unify(var, subst[val], subst)

    new_subst = subst.copy()
    new_subst[var] = val
    return new_subst

def apply_substitution(expr, subst):
    if is_variable(expr):
        return subst.get(expr, expr)
    if isinstance(expr, (list, tuple)):
        return [apply_substitution(item, subst) for item in expr]
    return expr


expr_a = ['Eats', 'x', 'Apple']
expr_b = ['Eats', 'Riya', 'y']

print(f"Expression A: {expr_a[0]}({expr_a[1]}, {expr_a[2]})")
print(f"Expression B: {expr_b[0]}({expr_b[1]}, {expr_b[2]})")


substitutions = unify(expr_a, expr_b)

if substitutions is not None:
    print("\nUnification Successful!")
    print(f"Substitutions: {substitutions}")


    unified_a = apply_substitution(expr_a, substitutions)
    unified_b = apply_substitution(expr_b, substitutions)

    print(f"Unified Expression A: {unified_a[0]}({unified_a[1]}, {unified_a[2]})")
    print(f"Unified Expression B: {unified_b[0]}({unified_b[1]}, {unified_b[2]})")
else:
    print("\nUnification Failed!")
```
