```python

from itertools import product

def implies(a, b):
    return (not a) or b


def knowledge_base(P, Q, R):
    return (
        implies(Q, P)
        and implies(P, not Q)
        and (Q or R)
    )


def tt_entails(query):
    # Enumerate every assignment to P, Q and R
    for P, Q, R in product([False, True], repeat=3):
        if knowledge_base(P, Q, R):
            # A counterexample disproves entailment
            if not query(P, Q, R):
                return False

    return True


def print_truth_table():
    print("P Q R | Q->P P->~Q Q|R | KB")

    for P, Q, R in product([False, True], repeat=3):
        a = implies(Q, P)
        b = implies(P, not Q)
        c = Q or R
        kb = a and b and c

        print(
            f"{P!s:5} {Q!s:5} {R!s:5} | "
            f"{a!s:5} {b!s:6} {c!s:4} | {kb}"
        )


print_truth_table()

print("KB entails R:", tt_entails(lambda P, Q, R: R))
print("KB entails R -> P:",
      tt_entails(lambda P, Q, R: implies(R, P)))
print("KB entails Q -> R:",
      tt_entails(lambda P, Q, R: implies(Q, R)))
```
