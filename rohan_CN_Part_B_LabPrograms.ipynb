```python

def min_val(x, y):
    if x < y:
        return x
    else:
        return y


drop = 0
count = 0

cap = int(input("Enter the bucket size:\n"))
process = int(input("Enter the processing rate:\n"))
nsec = int(input("Enter the number of seconds you want to simulate:\n"))

inp = []

for i in range(nsec):
    packet = int(input(f"Enter the size of the packet entering at {i + 1} sec:\n"))
    inp.append(packet)

print("\nSecond | Packet Received | Packet Sent | Packet Left | Dropped")
print("--------------------------------------------------------------")

for i in range(nsec):
    count += inp[i]

    if count > cap:
        drop = count - cap
        count = cap

    print(f"{i + 1}\t {inp[i]}\t\t", end="")

    mini = min_val(count, process)

    print(f"{mini}\t\t", end="")

    count = count - mini

    print(f"{count}\t\t {drop}")

    drop = 0

while count != 0:
    i += 1

    if count > cap:
        drop = count - cap
        count = cap

    print(f"{i + 1}\t 0\t\t", end="")

    mini = min_val(count, process)

    print(f"{mini}\t\t", end="")

    count = count - mini

    print(f"{count}\t\t {drop}")

    drop = 0
```

```python

import socket
import os
import threading

HOST = '127.0.0.1'
PORT = 65433

def server():
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as server_socket:
        server_socket.bind((HOST, PORT))
        server_socket.listen()

        print(f"Server listening on {HOST}:{PORT}")

        conn, addr = server_socket.accept()

        with conn:
            print(f"Connected by {addr}")

            filename = conn.recv(1024).decode()

            if os.path.isfile(filename):
                with open(filename, 'r') as f:
                    content = f.read()

                conn.sendall(content.encode())
            else:
                conn.sendall(b"File not found.")

thread = threading.Thread(target=server, daemon=True)
thread.start()
```

```python

import socket
import os
import threading

HOST = "127.0.0.1"
PORT = 5005

def server():
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)
    server_socket.bind((HOST, PORT))

    print(f"UDP Server listening on {HOST}:{PORT}")

    while True:
        filename, client_addr = server_socket.recvfrom(1024)
        filename = filename.decode()

        print(f"Client requested file: {filename}")

        if os.path.isfile(filename):
            with open(filename, "r") as f:
                data = f.read()

            server_socket.sendto(data.encode(), client_addr)
        else:
            server_socket.sendto(b"ERROR: File not found", client_addr)

thread = threading.Thread(target=server, daemon=True)
thread.start()
```
