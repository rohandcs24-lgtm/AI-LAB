```python
import math

def alpha_beta_pruning(depth, node_index, is_maximizing, values, alpha, beta, max_depth, pruned_nodes):

    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        best = -math.inf
        for i in range(2):
            val = alpha_beta_pruning(
                depth + 1, node_index * 2 + i, False, values, alpha, beta, max_depth, pruned_nodes
            )
            best = max(best, val)
            alpha = max(alpha, best)

            if beta <= alpha:

                for rem in range(i + 1, 2):
                    pruned_nodes.append((depth + 1, node_index * 2 + rem))
                break
        return best
    else:
        best = math.inf
        for i in range(2):
            val = alpha_beta_pruning(
                depth + 1, node_index * 2 + i, True, values, alpha, beta, max_depth, pruned_nodes
            )
            best = min(best, val)
            beta = min(beta, best)


            if beta <= alpha:
                for rem in range(i + 1, 2):
                    pruned_nodes.append((depth + 1, node_index * 2 + rem))
                break
        return best

if __name__ == "__main__":

    leaf_values = [10, 9, 14, 18, 5, 4, 50, 3]
    tree_depth = int(math.log2(len(leaf_values)))
    pruned = []

    optimal_value = alpha_beta_pruning(
        depth=0,
        node_index=0,
        is_maximizing=True,
        values=leaf_values,
        alpha=-math.inf,
        beta=math.inf,
        max_depth=tree_depth,
        pruned_nodes=pruned
    )

    print(f"Leaf Values: {leaf_values}")
    print(f"Final Value of Root MAX Node: {optimal_value}")
    print("Pruned (Depth, Node_Index):", pruned)
```
