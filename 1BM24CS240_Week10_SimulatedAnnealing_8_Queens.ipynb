```python
import random
import math

N = 8

def calculate_cost(state):

    attacks = 0
    for i in range(N):
        for j in range(i + 1, N):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_random_neighbour(state):

    neighbour = list(state)
    col = random.randint(0, N - 1)
    new_row = random.randint(0, N - 1)
    while new_row == neighbour[col]:
        new_row = random.randint(0, N - 1)
    neighbour[col] = new_row
    return neighbour

def simulated_annealing(initial_temp=1000.0, cooling_rate=0.995, min_temp=1e-5):

    current = [random.randint(0, N - 1) for _ in range(N)]
    current_cost = calculate_cost(current)
    T = initial_temp

    while T > min_temp and current_cost > 0:
        next_state = get_random_neighbour(current)
        next_cost = calculate_cost(next_state)


        delta_e = current_cost - next_cost

        if delta_e > 0:
            current = next_state
            current_cost = next_cost
        else:
            probability = math.exp(delta_e / T)
            if random.random() < probability:
                current = next_state
                current_cost = next_cost


        T *= cooling_rate

    return current, current_cost

def print_board(state):
    for row in range(N):
        line = ["Q" if state[col] == row else "." for col in range(N)]
        print(" ".join(line))

if __name__ == "__main__":

    solution, cost = simulated_annealing()
    attempts = 1
    while cost != 0:
        solution, cost = simulated_annealing()
        attempts += 1

    print(f"Solved 8-Queens in {attempts} run(s)!")
    print(f"State (Row per Column): {solution}")
    print(f"Final Attacking Pairs (Cost): {cost}\n")
    print_board(solution)
```
