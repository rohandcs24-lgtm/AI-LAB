```python

from itertools import combinations

def cost(state):
    attacks = 0

    for i, j in combinations(range(4), 2):

        if (state[i] == state[j] or
                abs(state[i] - state[j]) == abs(i - j)):
            attacks += 1

    return attacks


def hill_climbing():
    current = [3, 1, 2, 0]

    while True:
        current_cost = cost(current)
        print("Current:", current, "Cost:", current_cost)

        if current_cost == 0:
            print("Goal state reached!")
            return current

        neighbors = []

        for i, j in combinations(range(4), 2):
            neighbor = current.copy()
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append((cost(neighbor), i, j, neighbor))


        neighbors.sort(key=lambda item: (item[0], item[1], item[2]))

        best_cost, i, j, best_neighbor = neighbors[0]


        if best_cost >= current_cost:
            print("Stopped at a local minimum.")
            return current

        print(f"Swap columns ({i}, {j}) -> {best_neighbor}, "
              f"Cost = {best_cost}")
        current = best_neighbor


hill_climbing()
```
