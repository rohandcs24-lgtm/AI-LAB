# Forward Chaining Inference Engine

Forward chaining is one of the two main methods of reasoning when using an inference engine (the other is backward chaining). It starts with the available data and uses inference rules to extract more data until a goal is reached.

```python
class Rule:
    def __init__(self, premises, conclusion):
        self.premises = set(premises)
        self.conclusion = conclusion

    def __str__(self):
        return f"IF {', '.join(self.premises)} THEN {self.conclusion}"

def forward_chaining(rules, facts, goal):

    known_facts = set(facts)
    agenda = list(facts)
    inferred = {}

    count = {rule: len(rule.premises) for rule in rules}

    print("Initial Facts:", known_facts)
    print("Rules:")
    for rule in rules:
        print(f"  {rule}")
    print("\n--- Starting Inference ---")

    while agenda:
        p = agenda.pop(0)
        if p == goal:
            print(f"\nGoal '{goal}' has been reached!")
            return True

        if p not in inferred:
            inferred[p] = True
            for rule in rules:
                if p in rule.premises:
                    count[rule] -= 1
                    if count[rule] == 0:
                        new_fact = rule.conclusion
                        if new_fact not in known_facts:
                            print(f"Rule triggered! Derived new fact: {new_fact}")
                            known_facts.add(new_fact)
                            agenda.append(new_fact)

    print(f"\nGoal '{goal}' could not be reached. Derived facts: {known_facts}")
    return False

rules = [
    Rule(['A', 'B'], 'Y'),
    Rule(['C', 'Y'], 'Z'),
    Rule(['D'], 'C')
]

facts = ['A', 'B', 'D']
goal = 'Z'

forward_chaining(rules, facts, goal)
```
